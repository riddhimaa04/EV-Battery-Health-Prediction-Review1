```python

import os
import pandas as pd
import numpy as np

data_folder = "cleaned_dataset/data"

feature_rows = []

for _, row in discharge_data.iterrows():
    filename = row["filename"]
    file_path = os.path.join(data_folder, filename)

    try:
        battery_df = pd.read_csv(file_path)
        battery_df.columns = battery_df.columns.str.strip()

        # Skip files missing required measurement columns
        required_columns = [
            "Voltage_measured",
            "Current_measured",
            "Temperature_measured",
            "Current_load",
            "Voltage_load",
            "Time"
        ]

        if not all(col in battery_df.columns for col in required_columns):
            continue

        # Convert measurements to numbers
        for col in required_columns:
            battery_df[col] = pd.to_numeric(
                battery_df[col], errors="coerce"
            )

        # Remove rows without valid measurements
        battery_df = battery_df.dropna(subset=required_columns)

        if battery_df.empty:
            continue

        # Extract features from this discharge cycle
        features = {
            "filename": filename,
            "battery_id": row["battery_id"],
            "Capacity": row["Capacity"],
            "SOH": row["SOH"],
            "mean_voltage": battery_df["Voltage_measured"].mean(),
            "min_voltage": battery_df["Voltage_measured"].min(),
            "mean_current": battery_df["Current_measured"].mean(),
            "mean_temperature": battery_df["Temperature_measured"].mean(),
            "mean_load_current": battery_df["Current_load"].mean(),
            "mean_load_voltage": battery_df["Voltage_load"].mean(),
            "duration": (
                battery_df["Time"].max() -
                battery_df["Time"].min()
            )
        }

        feature_rows.append(features)

    except (OSError, ValueError, pd.errors.ParserError):
        continue

model_data = pd.DataFrame(feature_rows)

print("Dataset shape:", model_data.shape)
model_data.head()
```

```python
print("Dataset shape:", model_data.shape)
print("Missing values:")
print(model_data.isnull().sum())

print("\nSOH summary:")
print(model_data["SOH"].describe())
```

```python
model_data.columns = model_data.columns.str.strip()

print("Shape:", model_data.shape)
print("Columns:", model_data.columns.tolist())
print("Missing values:", model_data.isnull().sum().sum())
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

os.makedirs("outputs", exist_ok=True)

# Identify the SOH column
target_col = next(
    c for c in model_data.columns
    if c.lower() in ["soh", "sch"]
)

# 1. Distribution of battery health
plt.figure(figsize=(8, 5))
sns.histplot(model_data[target_col], bins=25, kde=True)
plt.title("Distribution of Battery State of Health")
plt.xlabel("SOH (%)")
plt.ylabel("Number of discharge cycles")
plt.tight_layout()
plt.savefig("outputs/soh_distribution.png", dpi=150)
plt.show()

# 2. Distribution of sensor measurements
sensor_cols = [
    "mean_voltage",
    "mean_current",
    "mean_temperature"
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for col, ax in zip(sensor_cols, axes):
    sns.histplot(model_data[col], kde=True, ax=ax)
    ax.set_title(col.replace("_", " ").title())

plt.tight_layout()
plt.savefig("outputs/sensor_distributions.png", dpi=150)
plt.show()

# 3. Correlation heatmap
corr_cols = [
    "mean_voltage", "min_voltage", "mean_current",
    "mean_temperature", "mean_load_current",
    "mean_load_voltage", "duration", target_col
]

plt.figure(figsize=(10, 7))
sns.heatmap(
    model_data[corr_cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("outputs/correlation_heatmap.png", dpi=150)
plt.show()

# 4. SOH trend for one battery
battery = model_data["battery_id"].iloc[0]
one_battery = model_data[
    model_data["battery_id"] == battery
].copy()

one_battery["cycle_number"] = pd.to_numeric(
    one_battery["filename"].str.extract(r"(\d+)")[0]
)

one_battery = one_battery.sort_values("cycle_number")

plt.figure(figsize=(9, 5))
sns.lineplot(
    data=one_battery,
    x="cycle_number",
    y=target_col,
    marker="o"
)
plt.title(f"SOH Trend for Battery {battery}")
plt.xlabel("Measurement file number (cycle proxy)")
plt.ylabel("Estimated SOH (%)")
plt.tight_layout()
plt.savefig("outputs/soh_trend.png", dpi=150)
plt.show()
```

```python
feature_cols = [
    "mean_voltage",
    "min_voltage",
    "mean_current",
    "mean_temperature",
    "mean_load_current",
    "mean_load_voltage",
    "duration"
]

# Show boxplots before outlier removal
plt.figure(figsize=(12, 6))
sns.boxplot(data=model_data[feature_cols])
plt.xticks(rotation=30)
plt.title("Outlier Detection in Sensor Features")
plt.tight_layout()
plt.savefig("outputs/outlier_boxplots_before.png", dpi=150)
plt.show()

# Calculate IQR limits
Q1 = model_data[feature_cols].quantile(0.25)
Q3 = model_data[feature_cols].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Count potential outliers
outlier_mask = (
    (model_data[feature_cols] < lower) |
    (model_data[feature_cols] > upper)
).any(axis=1)

print("Potential outlier rows:", outlier_mask.sum())
print("Rows before removal:", len(model_data))

# Remove rows flagged by the IQR rule
clean_data = model_data.loc[~outlier_mask].copy()

print("Rows after removal:", len(clean_data))
print("Rows removed:", len(model_data) - len(clean_data))
```

```python
encoded_battery_ids = pd.get_dummies(
    clean_data["battery_id"],
    prefix="battery",
    dtype=int
)

print(encoded_battery_ids.head())
print("Encoded columns:", encoded_battery_ids.shape[1])
```

```python
from sklearn.preprocessing import StandardScaler

X = clean_data[feature_cols].copy()
y = clean_data[target_col].copy()

scaler = StandardScaler()

X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=feature_cols,
    index=X.index
)

print("Standardized features:")
display(X_scaled.head())

print("\nFeature means (approximately zero):")
print(X_scaled.mean().round(3))

print("\nFeature standard deviations (approximately one):")
print(X_scaled.std(ddof=0).round(3))
```

```python
clean_data.to_csv(
    "processed_battery_data.csv",
    index=False
)

X_scaled.to_csv(
    "standardized_features.csv",
    index=False
)

print("Processed datasets saved successfully.")
```
